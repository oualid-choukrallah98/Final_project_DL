```python
import json
import matplotlib.pyplot as plt
import numpy as np

# Read the JSON files
with open('/workspace/Final_project_DL/lora/results/training_results_trial1_20251207_143640.json', 'r') as f:
    trial1 = json.load(f)

with open('/workspace/Final_project_DL/lora/results/training_results_trial2_20251207_170733.json', 'r') as f:
    trial2 = json.load(f)

# Extract losses
train_losses_trial1 = trial1['results']['train_losses']
val_losses_trial1 = trial1['results']['val_losses']
train_losses_trial2 = trial2['results']['train_losses']
val_losses_trial2 = trial2['results']['val_losses']

epochs_trial1 = list(range(1, len(train_losses_trial1) + 1))
epochs_trial2 = list(range(1, len(train_losses_trial2) + 1))

# Create the plot
plt.figure(figsize=(12, 6))

# Plot Trial 1
plt.plot(epochs_trial1, train_losses_trial1, 'b-', marker='o', label='Trial 1 - Train Loss', linewidth=2)
plt.plot(epochs_trial1, val_losses_trial1, 'b--', marker='s', label='Trial 1 - Val Loss', linewidth=2)

# Plot Trial 2
plt.plot(epochs_trial2, train_losses_trial2, 'r-', marker='o', label='Trial 2 - Train Loss', linewidth=2)
plt.plot(epochs_trial2, val_losses_trial2, 'r--', marker='s', label='Trial 2 - Val Loss', linewidth=2)

plt.xlabel('Epoch', fontsize=12)
plt.ylabel('Loss', fontsize=12)
plt.title('Training and Validation Losses (LoRA)', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.savefig('/workspace/Final_project_DL/lora/results/losses_plot.png', dpi=300, bbox_inches='tight')
plt.show()

# Print final results
print(f"\nTrial 1 - Final Train Loss: {train_losses_trial1[-1]:.4f}")
print(f"Trial 1 - Final Val Loss: {val_losses_trial1[-1]:.4f}")
print(f"\nTrial 2 - Final Train Loss: {train_losses_trial2[-1]:.4f}")
print(f"Trial 2 - Final Val Loss: {val_losses_trial2[-1]:.4f}")
```
